```python
import os
import random
import shutil

images_dir = "TrashDataset/images"
labels_dir = "TrashDataset/labels"

splits = ["train", "valid", "test"]

for split in splits:
    os.makedirs(f"data/{split}/images", exist_ok=True)
    os.makedirs(f"data/{split}/labels", exist_ok=True)

files = os.listdir(images_dir)

random.shuffle(files)

total = len(files)
train_end = int(0.8 * total)
valid_end = int(0.9 * total)

train_files = files[:train_end]
valid_files = files[train_end:valid_end]
test_files = files[valid_end:]

def copy_files(file_list, split):
    for file in file_list:
        #bild
        src_img = os.path.join(images_dir, file)
        dst_img = f"data/{split}/images/{file}"


        #label
        name = os.path.splitext(file)[0]
        src_lbl = f"{labels_dir}/{name}.txt"
        dst_lbl = f"data/{split}/labels/{name}.txt"

        # Kolla om label finns
        if not os.path.exists(src_lbl):
            print(f"Label saknas för {file} - skippar")
            continue


        shutil.copy(src_img, dst_img)
        shutil.copy(src_lbl, dst_lbl)

copy_files(train_files, "train")
copy_files(valid_files, "valid")
copy_files(test_files, "test")

print("Dataset split klart")
```

```python
missing = 0

for file in os.listdir(images_dir):
    name = os.path.splitext(file)[0]
    label = os.path.join(labels_dir, name + ".txt")

    if not os.path.exists(label):
        print("Saknar:", file)
        missing += 1

print("Totalt saknade labels:", missing)
```
