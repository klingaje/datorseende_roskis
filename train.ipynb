```python
from ultralytics import YOLO
import os

if not os.path.exists("data/train/images"):
    raise Exception("Hittar inte data/train/images")

if not os.path.exists("mydataset.yaml"):
    raise Exception("Hittar inte mydataset.yaml")

print("Dataset hittat ✓")
print("Startar träning...")

model = YOLO("yolo26n.pt")

results = model.train(
    data = "mydataset.yaml",
    epochs = 50,
    patience = 15,
    flipud = 0.2,
    fliplr = 0.5,
    mixup = 0.1,
    copy_paste = 0.1,
    degrees = 15.0,
    shear = 5.0,
    imgsz = 640,
    batch = 8,
    name = "roskispolis"
)

print("Träning klar!")
print("Modell sparad i:")
print("runs/detect/roskispolis/")
```
